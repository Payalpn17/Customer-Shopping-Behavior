```python
import sqlite3
from sqlalchemy import create_engine

# 1. Create a local SQLite database connection
engine = create_engine("sqlite:///portfolio.db")

# 2. Export your cleaned DataFrame to a SQL table named 'customer_shopping'
df.to_sql("customer_shopping", engine, if_exists="replace", index=False)

# 3. Query your table using SQL
query = "SELECT category, AVG(purchase_amount) AS avg_spent FROM customer_shopping GROUP BY category"
sql_results = pd.read_sql(query, engine)

sql_results
```

```python
# 1. Rename the column to remove the special characters
df = df.rename(columns={'ï»¿customer_id': 'customer_id'})

# 2. Re-export the DataFrame to overwrite the SQLite table
df.to_sql("customer_shopping", engine, if_exists="replace", index=False)

# 3. Run your SQL query again
query = """
SELECT customer_id, purchase_amount
FROM customer_shopping
WHERE discount_applied = 'Yes'
  AND purchase_amount >= (SELECT AVG(purchase_amount) FROM customer_shopping)
"""

sql_results = pd.read_sql(query, engine)
sql_results
```

```python
query="select item_purchased, round(avg(review_rating),2) as average_rating from customer_shopping group by item_purchased order by avg(review_rating) desc  limit 5 "
sql_results = pd.read_sql(query, engine)
sql_results
```

```python
query="select shipping_type, round(avg(purchase_amount),2) as Average_Purchase_Amount from customer_shopping where shipping_type in ('Standard', 'Express') group by shipping_type"
sql_results = pd.read_sql(query, engine)
sql_results
```

```python
query="select subscription_status, count(customer_id) as total_customers, round(avg(purchase_amount),2) as Spend, round(sum(purchase_amount),2) as Revenue  from customer_shopping group by subscription_status order by Revenue,Spend"
sql_results = pd.read_sql(query, engine)
sql_results
```

```python
query=" select item_purchased, round(100 * sum(case when discount_applied='Yes' then 1 else 0 end)/count(*),2) as discount_rate from customer_shopping group by item_purchased order by discount_rate desc limit 5"
sql_results = pd.read_sql(query, engine)
sql_results
```

```python
query=""" with customer_type as (
         select customer_id, previous_purchases,
         case when previous_purchases=1 then 'New'
              when previous_purchases between 2 and 10 then 'Returning'
              else 'Loyal'
          end as customer_segment
        from customer_shopping
        )
      select customer_segment,count(*) as Number_of_customers
      from customer_type
      group by customer_segment """
sql_results = pd.read_sql(query, engine)
sql_results
```

```python
query="""
    with item_count as (
    select category, item_purchased,count(customer_id) as total_orders,
    row_number() over (partition by category order by count(customer_id) desc) as item_rank
    from customer_shopping
    group by category,item_purchased
    )
    select item_rank,category,item_purchased,total_orders
    from item_count
    where item_rank<=3  """
sql_results = pd.read_sql(query, engine)
sql_results
```

```python
query="""
      select subscription_status,count(customer_id) as repeat_buyers
      from customer_shopping
      where previous_purchases>5
      group by subscription_status
      """
sql_results = pd.read_sql(query, engine)
sql_results
```
